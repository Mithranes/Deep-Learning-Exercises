Resizing the image using Pillow library

```python
from PIL import Image
```

```python
img = Image.open('Dataset/dog.jpeg')
img_resized = img.resize((200, 200))
```

```python
img_resized.save('dog_resized.jpeg')
```

```python
img_resized = mpimg.imread('Dataset/dog_resized.jpeg')
```

```python
plt.imshow(img_resized)
plt.show()
```

```python
img_resized.shape
```

Converting RGB to Graysacle using OpenCV

```python
# importing OpenCV Library
import cv2
```

```python
img = cv2.imread('Dataset/dog.jpeg')
```

```python
type(img)
```

```python
img.shape
```

```python
grayscale_image = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
```

```python
type(grayscale_image)
```

```python
grayscale_image.shape
```

```python
cv2.imshow("Grayscale Dog", grayscale_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
plt.imshow(grayscale_image)
plt.show()
```

```python
# Saving grayscale image
cv2.imwrite('dog_grayscale.jpeg', grayscale_image)
```
